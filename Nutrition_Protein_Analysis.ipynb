```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('/content/Food_Nutrition_Dataset.csv')

# Clean data - remove missing protein values
df = df.dropna(subset=['protein'])

# Top 10 protein-rich foods
top_protein = df.sort_values(by='protein', ascending=False).head(10)
print(top_protein[['food_name', 'protein']])

# Visualization
plt.figure(figsize=(10,6))
plt.barh(top_protein['food_name'], top_protein['protein'], color='teal')
plt.xlabel('Protein (g)')
plt.title('Top 10 Protein-Rich Foods')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Correlation: Protein vs Calories
correlation = df['protein'].corr(df['calories'])
print(f"Correlation between Protein and Calories: {correlation:.2f}")
```
